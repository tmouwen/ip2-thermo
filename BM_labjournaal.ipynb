```python
#Data processing and analysis:
from scipy.optimize import curve_fit
U=R3/(R2+R3)*u_bron
U_D= U-UR1
I_D=UR1/R1

plt.plot(U_D, I_D, '.r')
plt.semilogy()
plt.grid()
plt.xlabel('Ud')
plt.ylabel('Id')
q = -1.602e-19
n = 2
T = 294
U_D_min = 0.1
mask = (U_D > U_D_min) & (I_D > 0)
U_D_fit = U_D[mask]
I_D_fit = I_D[mask]
def function(V_D,log10_I0,b):
    I0 = 10**log10_I0
    return np.log10(I0*(np.exp(b*V_D)-1))





val, cov = curve_fit(function,U_D_fit,np.log10(I_D_fit),p0=[-9, 20],bounds=([-20, 1], [0, 100]))

print( val)


U_lijn = np.linspace(min(U_D_fit), max(U_D_fit), 500)


I_lijn = 10**function(U_lijn, *val)

# Grafiek
plt.figure()
plt.semilogy(U_D, I_D, 'ro', label='Meetwaarden')
plt.semilogy(U_lijn, I_lijn, 'b-', label='Fit')
plt.xlabel('Ud (V)')
plt.ylabel('Id (A)')
plt.grid()
plt.legend()
plt.show()
```
